```python
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# Extract variables
lon = da["x"].values
lat = da["y"].values
siconc = da["avg_siconc"].values

# Create a plot with a geographic projection
fig, ax = plt.subplots(figsize=(8, 6), subplot_kw={'projection': ccrs.PlateCarree()})

# Scatter plot of the data
sc = ax.scatter(lon, lat, c=siconc, cmap='coolwarm', s=10, edgecolors='none', alpha=0.75, transform=ccrs.PlateCarree())

# Add coastline
ax.coastlines()
ax.add_feature(cfeature.BORDERS, linestyle=':', edgecolor='black')

# Add gridlines
ax.gridlines(draw_labels=True)

# Colorbar
cbar = plt.colorbar(sc, ax=ax, orientation="vertical")
cbar.set_label('Sea Ice Concentration (Fraction)')

# Labels and title
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
ax.set_title("Sea Ice Concentration with Coastline")

plt.show()
```
